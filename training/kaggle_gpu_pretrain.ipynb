# Lowcar: GPU pretraining for the chase brains

Trains the learned cops and runner in the GPU copy of the chase (a faithful port of the game's car physics,
crashes, senses and rules), many times faster than the real game. The brains then continue in the real game
with the GitHub chase training.

**Before running:** in the notebook settings (right panel) set **Accelerator: GPU T4 x2** (or P100) and
turn **Internet on**. Then use **Save Version → Save & Run All (Commit)** so it keeps running in the background
(up to 12 hours) even with this tab closed. The finished brains appear under **Output** (`out/`).
